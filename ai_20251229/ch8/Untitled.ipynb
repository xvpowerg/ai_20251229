## 今日作業
如果甚麼都沒輸入 直接按Enter就結束程式
如果有輸入判斷輸入的水果是否在List中

請輸入水果(Enter結束)橘子

橘子在清單

請輸入水果(Enter結束)櫻桃

櫻桃不在清單內
內

```python
fruits = ["香蕉","蘋果","橘子","西瓜"]

while True:
    myFruit = input("請輸入水果(Enter結束)")
    if myFruit == "":
        break
    if myFruit in fruits:
        print(f"{myFruit}在清單")
    else:
        print(f"{myFruit}不在清單")
```

## 新增與修改
+ 新增一筆資料
  + 請將 key 與 value 儲存至一個變數內。
  + 透過 dict.update(新增的資料) 這個方法的方式就可以新增。
+ 元素可否變更：
  + 以 = 指派方式指派給 key 就可以變更資料

## update

```python
dict2 = {"ken":100,"vivin":95,"lucy":72}
print(dict2)
dict2.update({"vivin":63})
print(dict2)
#批次改
dict2.update({"ken":98,"lucy":100})
print(dict2)
#新增找不到key就新增
dict2.update({"iris":68,"vivin":96,"tom":77})
print(dict2)
```

```python
dict3 = {"ken":100,"vivin":95,"lucy":72}
print(dict3)
dict3["ken"] = 88
print(dict3)
dict3["iris"] = "Score"
print(dict3)
```

```python
k = "tom"
if k in dict3:
    print(dict3[k])
else:
    print(f"{k}不存在")
```

## del 

del dict[key]    刪除某一個 key 的資料

dict.clear( )     清除所有項目

del dict           刪除字典

```python
dict4 = {"ken":100,"vivin":95,"lucy":72}
print(dict4)
del dict4["vivin"]
print(dict4)
```

```python
dict4 = {"ken":100,"vivin":95,"lucy":72}
#key存在才能刪除 
#key不存在不能刪
#不希望產生錯誤
k = "vivin"
if k in dict4:
    del dict4[k]
else:
    print(f"{k}不存在無法刪除")
print(dict4)
```

## 自訂函式的運用

#### 程式設計時儘量減少相同語法的重複出現，為什麼?

修改時可能忘記其他地方沒有改，導致程式執行出現錯誤結果。

減少重複語法對於之後的維修會比較方便。


#### 要如何減少重複的相同語法?

把重複的相同語法當作工具看待，需要的時候呼叫他。

這種工具於程式語法內叫做函數(function)。


```python
def func_sum():
    print("呼叫函式")
    
func_sum()#呼叫函式
func_sum()
func_sum()
```


### 自訂函式接收資料：

傳入函式的參數放在 ( ) 內。

可以接收多個，以逗點隔開。

接收參數可以是變數，也可以是 list。

函式若設定接收參數，呼叫函式時一定要給參數。

接收參數可設定為不固定數量。

若不接收則為空白。
若不接收則為空白。


```python
def func_sum(a,b):
    c = a + b
    print(c)
func_sum(2,3)
```

```python
def forLoop(a):
    for i in range(a):
        print(i)
    print("========") #我是分隔線   
forLoop(3)
forLoop(5)
```

```python
def test1(a,b):
    c = a + b
    print(c)
    
def test2(x):
    ans = x ** 2
    print(ans)
test1(2,3)
test2(6)
```

```python
v1 = test1(2,3)#目前沒有回傳值
print(v1)#None表示空
```

```python
def test1(a,b):
    c = a + b
    return c #定義回傳值
    
v1  = test1(3,6)
print(v1)
```

```python
def test3(a,b):
    c = a * b
    return c
def test4(a,b):
    c = a - b
    return c
    
x1 = test3(2,3)
x2 = test3(5,2)
ans2 = test4(x1,x2)
ans = test4(test3(2,3),test3(5,2))
print(ans)
print(ans2)
```

```python
def test4(a,b,x):
    c = x
    print(a,b,c)
#6,7,9
test4(6,7,10)    
```

## 變數影響範圍
+ 函數外的變數：
  + 函數內可以顯示該變數內容
  + 不屬於函數的區域內都可以使用
+ 函數內的變數：
  + 只在函數內產生效果，不會影響函數外的變數
  + 若函數內沒有進行變數宣告而進行改變內容動作將會產生錯誤訊息!

```python
def func_sum():
    a = 10
    print("函數內:",a)
a = 5
print("函數外1:",a)
func_sum()
print("函數外2:",a)
```

```python
def checkFunc():
    print("內部:i",i)
i = 10
print("外部:i",i)
checkFunc()
```
