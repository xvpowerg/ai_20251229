```python
a,b = 10,3
print(a,"+",b,"=",a + b)
ans = a - b
print(a,"-",b,"=",ans)
ans = a * b
print(a,"*",b,"=",ans)
ans = a / b
#round(ans,2) 取小數點第2位
print(a,"/",b,"=",round(ans,2) )
ans = a // b
print(a,"//",b,"=",ans)
ans = a % b #餘數
print(a,"%",b,"=",ans)
ans = a ** 2
print(a,"2次方=",ans)
```

```python
##關係運算子
a = 10
b = 5
print(a,">",b,"=",a > b)
print(a,"<",b,"=",a < b)
print(a,">=",b,"=",a >= b)
print(a,"<=",b,"=",a <= b)

print(a,"==",b,"=",a == b)
print(a,"!=",b,"=",a != b)#不等於
```
