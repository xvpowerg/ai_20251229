### 使用 loc(文字 m:n 會包含到n)
+ "location" 的縮寫
+ .loc[ ], 中括弧裡面是先 row 後 column，以逗號分開。
+ row 與 column 要放入的是 row 標籤與 column 標籤。
+ 可於 row 或 column 加入「：」 ，代表指定起訖範圍。

```python
import  pandas as pd
import  numpy as np
df=pd.DataFrame(
    np.arange(48).reshape(6,8),
    index=['a','b','c','d','e','f'],
    columns=['s','t','u','v','w','x','y','z'])
print(df)
print()
print(df.loc["b","v"])#左邊表示軸0 右邊表示軸1
print("="*20)
t1 = df.loc["b",]
print(t1,type(t1))
print("="*20)
t1 = df.loc["b","u":"x"]#會包含x
print(t1,type(t1))
print("="*20)
t2 = df.loc[['a','c'],['t','v','x']]
print(t2,type(t2))
print("="*20)
t3 = df.loc[["b"],]#DataFrame
print(t3,type(t3))
```

### 使用 iloc
+ iloc[ ] 與 loc 一樣，中括弧裡面也是先 row 後 column，行列標籤用逗號分開。
+ 與 loc 不同的之處是，.iloc 是根據索引來進行選擇。

```python
import  pandas as pd
import  numpy as np
df=pd.DataFrame(
    np.arange(48).reshape(6,8))
print(df)
print("========")
print(df.iloc[1,3:6])#不包含6
print("========")
print(df.iloc[2,3])
print("========")
print(df.iloc[:,-4])
print("========")
print(df.iloc[:,-4:])
```
