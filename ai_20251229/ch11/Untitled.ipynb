## 正規表示式



```python
import re
s = "Tim's phone numbers are 12345-41521 and 78963-85214"
match = re.findall(r"\D",s)
print(match)
```

```python
import re
s = "Tim's phone numbers are 12345-41521 and 78963-85214"
match = re.findall(r"\d{3}",s)
print(match)
match = re.findall(r"\d{1,3}",s)
print(match)
```

```python
import  re
s="Python was conceived in the late 1980s by Guido van Rossum"
match = re.findall(r"[a-t]",s)
print(match)
match = re.findall(r"[a-t]{2}",s)
print(match)
match = re.findall(r"[a-t]{3,6}",s)
print(match)
```

```python
import re
pattern = re.compile(r"\d+")

r1 = pattern.findall("1hello 123456 789")
print(r1)
```

## split 方法
+ split 方法能以比對規則針對字串進行切割後產生 list。
+ split 它的使用形式如下：
+ split(字串,最大分割次數)
   + 若不指定最大分割次數則全部分割

```python
import re
text = 'one, two...ten'
re1 = re.split("[,. ]+",text)
print(re1)
```

## sub 方法
+ sub 方法用於替換，它的使用形式如下
  + sub(取代的字串或函數,來源字串,可設定被取代的數量)
+ 可設定被取代的數量若不指定時全部替換。

```python
import re
inputStr = "hello 123 world 456"
rStr = re.sub(r"\d+","222",inputStr)
print(rStr)
```

## 物件導向入門

```python
class Employee: #屬性
    pass#跳過不要出錯
emp = Employee()#物件
print(emp)
emp.name = "Howard"#物件.屬性名稱 = xxxx 建立屬性
emp.salary = 50000
print(emp.name,emp.salary)
```

```python
class Employee:
    #建構式
    def __init__(self,name,salary=28000):
        print("self:",self)
        self.name = name
        self.salary = salary

emp1 = Employee("Ken",38500)        
print("emp1:",emp1)        
print("emp1:",emp1.name,emp1.salary)        
emp2 = Employee("Joy")
print("emp2:",emp2.name,emp2.salary)      
```
