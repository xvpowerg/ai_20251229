### Del
+ list 可透過 del 方式移除指定位置資料。
+ del list[ ] 必須指定範圍：
  + del list[n] 代表移除第 n 位索引值資料。
  + del list[:m] 代表移除第 m 位索引值之前 ( 不包含 m ) 資料。
  + del list[n:m] 代表移除第 n 到 m 位索引值之前 ( 不包含 m ) 資料。

```python
list2 = ["A","B","C","D","E"]
print(list2)
del list2[1]
print(list2)
print(list2[1:3])
del list2[1:3]
print(list2)
```

## Dictionary

資料結構是由 key:value 所組成。

key 不能夠重複，否則會被後面的結果蓋過去。

可輸入 key 找尋您要找出來的值。

如果輸入的 key 不存在，那就會出現錯誤訊息。

key 只能是 字串 數字 tuple

```python
ditct1 = {"A":100,"B":200,"C":300}
print(ditct1)
print(ditct1["B"])
print(ditct1["A"])
print(ditct1["X"])# KeyError 因為X的key不存在所以出錯
```

```python
dict2 = {"a":100,"b":200,"b":300}
print(dict2)
```

```python

dict3 = {"a":100,"b":200,"c":300}
for key in dict3:
    print(key,dict3[key])
```

```python
dict3 = {"a":100,"b":200,"c":300}
print(dict3.values())
for value in dict3.values():
    print(value)
```
