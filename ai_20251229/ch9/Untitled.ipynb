## 作業 寫一個函式myMax 傳2個參數 a , b幫我回傳最大值 

```python
def myMax(a,b):
    if a > b:
        return a
    else:
        return b
m1 = myMax(10,2) 
print(m1)
m2 = myMax(5,8)
print(m2)
```

### 不固定接收數量

( ) 內接收資料若是以 * 表示代表可以引入不定數量的參數：

一個*代表以 tuple 的方式引入

兩個*代表以 dict 的方式引入

接收資料若需加入多個設定，這些設定請依照順序排列：

第一個參數設定為接收值 

第二個參數設定為初始值

第三個參數設定為不固定值，可以是 tuple 或 dict 方式

```python
def printInfo(age1,age2="default",*age3):
    print(age1,age2,age3)
printInfo(1)
printInfo(1,2)
printInfo(1,2,3)
printInfo(1,2,3,4,5,6)
```

```python
def printInfo(*age3,age1,age2="default"):
    print(age1,age2,age3)
printInfo(1,2,3,4,5,age2=" Joy",age1="Hello")#只要使用命名參數 接下來的參數都要用    命名參數 
```

```python
def fun4(arg1,arg2 = "default",**arg3):
    print(arg1)
    print(arg2)
    print(arg3)
    print("=======================")
fun4(1)
fun4(1,2)
fun4(1,2,a="Ken",score=80)
```

```python
def sumValue(*arg):
    v = 0
    for a in arg:
        v += a
    return v
print(sumValue())
# print(sumValue(2))
print(sumValue(2,3))
print(sumValue(2,3,4))
print(sumValue(2,3,4,5))
```

## 匿名函數
不需要定義函數名稱，只需要用到運算式或表達分析語法。

Python使用lambda語法定義匿名函數。

匿名函數是一個表達式或計算式，並不是一個執行流程區塊。

匿名函數出現在一般函數不允許的地方，例如list裡面或著函數呼叫參數的位置。


```python
def test1(a1):
    print(a1)
test1("Hello!")
f1 = test1
f1("Joy")
```

```python
def test2(a,b,call):
    c = call(a,b)
    print("ans:",c)
def func1(x1,x2):
    return x1 * x2
def func2(x1,x2):
    return x1 - x2
test2(2,5,func1)  
test2(2,5,func2)  
```

```python
def test2(a,b,call):
    c = call(a,b)
    print("ans:",c)
f1 = lambda x1,x2: x1 * x2
f2 = lambda x1,x2 : x1 - x2
test2(2,5,f1)
test2(2,5,f2)
#某種語句
test2(2,3,lambda x1,n:x1**n)
```

```python
def f(x,y,z):
    return x +y +z
print(f(2,30,400))    
f2 = lambda x,y,z:x+y+z
print(f2(2,30,400))
```
