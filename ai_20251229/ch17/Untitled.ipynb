```python
import  matplotlib.pyplot as plt
import  numpy as np
city=['Delphi','Beijing','Washington','Tokyo','Moscow']
pos=np.arange(len(city))
gender=['Male','Female']

happing_index_female=[30,60,70,55,75]
happing_index_male=[60,40,70,65,85]
plt.bar(pos,happing_index_male,color = "blue")
plt.bar(pos,happing_index_female,color = "pink")

plt.xticks(pos,city) 
plt.xlabel('capital')
plt.ylabel('index',rotation=0)
plt.legend(gender,loc=5)#小圖示
plt.title('barchart')
plt.show()
```

```python
import  matplotlib.pyplot as plt
import  numpy as np
city=['Delphi','Beijing','Washington','Tokyo','Moscow']
pos=np.arange(len(city))
gender=['Male','Female']

happing_index_female=[30,60,70,55,75]
happing_index_male=[60,40,70,65,85]
plt.bar(pos,happing_index_male,color = "blue",bottom=happing_index_female)
plt.bar(pos,happing_index_female,color = "pink")

plt.xticks(pos,city) 
plt.xlabel('capital')
plt.ylabel('index',rotation=0)
plt.legend(gender,loc=2)#小圖示
plt.title('barchart')
plt.show()
```

```python
import  matplotlib.pyplot as plt
import  numpy as np
city=['Delphi','Beijing','Washington','Tokyo','Moscow']
pos=np.arange(len(city))
gender=['Male','Female']

bar_width = 0.35
happing_index_female=[30,60,70,55,75]
happing_index_male=[60,40,70,65,85]
plt.bar(pos,happing_index_male,bar_width,color = "blue")
plt.bar(pos+bar_width,happing_index_female,bar_width,color = "pink")

plt.xticks(pos,city) 
plt.xlabel('capital')
plt.ylabel('index',rotation=0)
plt.legend(gender,loc=2)#小圖示
plt.title('barchart')
plt.show()
```

## 合併

```python
import pandas as pd
d = {'one':pd.Series([1,2,3],index=['a','b','c']),
     'two':pd.Series([1,2,3,4],index=['a','b','c','d'])}
df1 = pd.DataFrame(d)
df1["three"] = [10,11,12,13]
print(df1)
df2 = pd.DataFrame([[20, 21, 22],[23, 24, 25],[26,27, 28]], columns=['one', 'two', 'three'],index=['a','b','c'])
print(df2)
df3 = pd.concat([df1,df2],ignore_index = True)#預設使用axis=0合併 忽略原本的index ignore_index = True
df3.index = ["a","b","c","d","e","f","g"]
print("="*20)
print(df3)
df4 = pd.concat([df1,df2],ignore_index = True,axis=1)
print(df4)
```
