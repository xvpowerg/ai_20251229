```python
#建立一個Student 類別 我希望可以 有兩個屬性 name score
#st1 = Student("Ken",85)
#st2 = Student("Iris",73)
#print(st1) Ken 85
#print(st2) Iris 73
#print(st1 < st2) False 比成績
```

```python
class Student:
    def __init__(self,name,score = 0):
        self.name = name
        self.score = score
    def __str__(self):
        return f"{self.name} {self.score}"
    def __lt__(self,other):    
        return self.score < other.score
st1 = Student("Ken",85)
st2 = Student("Vivin",91)         
st3 = Student("Lindy",71)         
st4 = Student("Joy",83)  
print(st1 < st2)
myList = [st1,st2,st3,st4]
myList.sort()
for s in myList:
    print(s)
```

## Json





+ 以 json.dumps( ) 函數從 Python 物件轉入 轉出JSON格式字串。
+ 以 json.dump( )函數從 Python 物件轉入 轉出JSON 檔案中。
+ json資料於Python處理UTF8碼內容會產生亂碼，建議 dumps 時加入以下的參數才可以正確處理UTF8碼內容內容

```python
import json
json1 = {"python":"課程","gjun":100,"python-class":True,"Line":None}
print(json1)
print(type(json1))
jsonStr = json.dumps(json1,ensure_ascii=False)
print(jsonStr)
```

```python
import json
json2 = {"python":"課程","gjun":100,"python-class":True,"Line":None}
with open("data0.json","w",encoding="utf-8") as outfile:
    json.dump(json2,outfile,ensure_ascii=False)
```

```python
import json
data = { }
data['people'] = [ ]
data['people'].append({
    'name': 'Scott',
    'website': 'stackabuse.com',
    'from': 'Nebraska'
})
data['people'].append({
    'name': 'Larry',
    'website': 'google.com',
    'from': 'Michigan'
})
print(data)
with open("data.json","w",encoding="utf-8") as outfile:
    json.dump(data,outfile,ensure_ascii=False,indent=1)
```

### Json 轉為 Python物件
+ 以 json.loads( ) 函數從 JSON 字串中取出資料轉入 Python。
+ 以 json.load( ) 函數從 JSON 檔案中取出資料轉入 Python。


```python
import json
json1 = """ 
{
    "python":"good",
    "gjun":100,
    "python-class":true
}
"""
print(json1)
json2 = json.loads(json1)
print(json2)
```

```python
import json
with open("data.json","r",encoding="utf-8") as jsonFile:
    data = json.load(jsonFile)
    for k in data:
        print(f"{k}-{data[k]}")
```

```python
from bs4 import BeautifulSoup
html='''<a id='a1' class="c1" href="https://google.com">Google</a>
<a id='a2' class="c1" href="https://yahoo.com">Yahoo</a>
<a id='a3' class="c3" href="https://facebook.com">Facebook</a>
'''
soup = BeautifulSoup(html,"html5lib")
print(soup.find(id="a1").getText())
print(soup.find(id="a1").get("href"))
tmpList = soup.find_all(class_="c1")
for tag in tmpList:
    print(tag.get("href"),tag.getText())
```
