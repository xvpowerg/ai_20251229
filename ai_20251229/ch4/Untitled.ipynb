```python
h = float(input("請輸入身高單位為公分"))
w = int(input("請輸入體重單位公斤"))
bmi = w / (h / 100) ** 2
print(bmi)
if   0 < bmi <= 18.5:
    print("太輕")
elif bmi > 18.5 and bmi <= 25:
    print("正常")
elif 25 < bmi <= 30:
    print("過重")
else:
    print("胖胖")
```

## break 代表不回頭，跳出迴圈。
## continue 代表繼續執行下一個迴圈。

```python
#for 寫一個輸出1 ~ 10
x = 3
for i in range(1,11):
    print(i,end = " ")
    if i == x:
        break#離開迴圈
```

```python
for i in range(1,11):
    print(i,end=" ")
    if i == 3 or i == 5 or i == 7:
        print()
        continue
    print("Hello:",i)
```

```python
print("請輸入大於80的整數")
while True:
    a = int(input("請輸入數字:"))
    if a > 80:
        print("a=",a)
        break
```
