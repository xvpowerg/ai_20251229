# List VS Tuple
+ tuple 與 list 的差異
  + 括號不同：
    + tuple 是用小括號將元素包起來。
    + list 是用中括號將元素包起來。
+ 元素可否變更：
   + tuple 決定了元素的內容就不可以變更。
   + list 可以隨時修改

```python
myList = ["Ken","Joy","Lucy"]
myTuple = ("Iris","Tom")
print(myList)
print(myTuple)
```

```python
#修改
print(myList[0])
print(myTuple[0])
myList[0] = "Lindy"
print(myList)
```

```python
myTuple[0] = "XXX" #Tuple只能讀 不修改刪除新增
```

```python
myList = ["Ken","Joy","Lucy","Tom"]
msg = ""
#index 索引 是List的編號 由0開始
#index 最大值是len - 1
print(len(myList))#長度 
print("index最大值是:",len(myList) - 1)

i = 0
print(myList[i])
i = 1
print(myList[i])
i = 2
print(myList[i])
i = 3
print(myList[i])
print("==========================================")#我是分隔線沒有用
for i in range(len(myList)):
    print(i,myList[i])
    msg += myList[i]
    if i < len(myList) - 1:
        msg += ","
#i = 4
#print(myList[i])

print(msg)#Ken,Joy,Lucy,Tom
```

```python
myList = ["Ken","Joy","Lucy","Tom"]
msg = ",".join(myList) #List必須全都為字串類型
#我要使用,將List的內容 使用 join合再一起
print(msg)
```

## 作業 寫一隻程式判斷是否為迴文

```python
str1 = "ABA" #str1是迴文 左往右讀 或 右往左讀都一樣稱迴文
str2 = "CAA"#str2不是迴文
print(str1[::-1])
# 寫一隻程式判斷是否為迴文
#是迴文顯示是迴文
#不是迴文顯示不是迴文
```
