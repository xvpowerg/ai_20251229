## list 項目插入新資料
+ list 可使用 insert 方式插入新的資料：
   + list.insert(索引值，插入的資料)。
   + 若索引值超出範圍則會改為 append 方式附加。

```python
myList = ["a","b","c","d","e"]
print(myList)
myList.insert(1,256)
print(myList)
myList.insert(-2,128)
print(myList)
myList.append(999)
print(myList)
myList.insert(0,777)
print(myList)
```

## 移除 
+ list 可透過以下三種方式移除項目：
  + list.remove( )
  + list.pop( )
  + del list[ ]

###  remove

依據內容進行刪除

list.remove(項目內容)

```python
myList1 = ["1","x","2","x","3"]
myList1.remove("x") #左往右找到第一個x移除
print(myList1)
myList1.remove("x")
print(myList1)
n = "3"
if n in myList1:
    myList1.remove(n)
print(myList1)
```

```python
myList1 = ["1","x","2","x","3"]
n = "x"
while True:
    if n in myList1:
        myList1.remove(n)
    else:
        break
print(myList1)
```

```python
myList1 = ["1","x","2","x","3"]
n = "x"
while True:
    if n not in myList1: #n不在myList1 做甚麼
        break
    myList1.remove(n)  
print(myList1)
```

### pop

如果 pop( ) 內沒有參數，則移除最後一筆資料。

pop() 內若有參數，則移除指定位置的資料。

```python
myList2 = ["Ken"]
myList2.append("Vivin")
myList2.append("Lindy")
myList2.append("Lucy")
print(myList2)
n1 = myList2.pop()#預設情況下移除最後一筆
print(n1)
print(myList2)
n2 = myList2.pop(1)
print(n2)
print(myList2)
n3 = myList2.pop()
print(n3)
print(myList2)
n4 = myList2.pop()
print(myList2)
```
