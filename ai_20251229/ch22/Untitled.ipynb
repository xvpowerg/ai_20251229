## 線性迴歸訓練預估

```python
import pandas as pd
tvmarketing = pd.read_csv("tvmarketing.csv")
print(tvmarketing.head())
```

```python
print(tvmarketing.isnull().any())
#如果有一個空值顯示True
#如果有一個空值數量
print(tvmarketing.isnull().sum())
```

```python
print(tvmarketing.dtypes)
```

```python
import matplotlib.pyplot as plt
plt.scatter(tvmarketing["TV"],tvmarketing["Sales"])
plt.xlabel("TV")
plt.ylabel("Sales")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
import sklearn.linear_model
X = tvmarketing["TV"]
y = tvmarketing["Sales"]
print('準備做資料切割')
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3)# 70% 做訓練 30%做測試
print(X_train.shape)
print(X_test.shape)
```

```python
model = sklearn.linear_model.LinearRegression()
import numpy as np
print(X_train.shape)
X_train = np.array(X_train).reshape(-1,1)
print(X_train.shape)
model.fit(X_train,y_train)
```

```python
print("係數:",model.coef_)
print("截距:",model.intercept_)
#y = ax + b
```

```python
import matplotlib.pyplot as plt
X2=np.array(X).reshape(1,-1)
y2 = (model.coef_)*X2+model.intercept_
plt.scatter(X2,y)
plt.scatter(X2,y2,c="red")
plt.show()
```

```python
import numpy as np
a = [[15,14,15],
     [12,13,17]]
b = [[10,14],
    [5,18],
    [8,10]]
a = np.array(a)
b = np.array(b)
ashape = a.shape
bshape = b.shape
print(ashape)
print(bshape)
#判斷是否能相乘
if ashape[1] == bshape[0]:
    print("可以相乘")
    ans = a @ b
    print(ans)
    print(ans.shape)
else:
    print("無法相乘")
    
```
