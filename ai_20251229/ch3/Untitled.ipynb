## FOR迴圈
### 知道數量時使用for loop
### python的for 有一種由頭到取數值的


想法

```python
r1 = range(5)#0 1 2 3 4
print(r1)#range 只會定義 數字數量 不會真的產生數字
print(list(r1))
```

```python
v1 = 1
v2 = 6
r2 = range(v1,v2)#>= v1  < v2
print(list(r2))
```

```python
r3 = range(0,5)
print(list(r3))
```

```python
r4 = range(1,11)
print(list(r4))
```

```python
r5 = range(1,11,2)
print(list(r5))
```

```python
for i in range(5):
    print(i)
```

```python
for k in  range(2,8):
    print(k,end=" ")
```

```python
print("A",end="@")#end表示 print的結尾要輸出什麼
print("B",end="!")
print("C",end="*")
```
