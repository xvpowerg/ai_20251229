### in 與 not in

+ Pandas 有 isin 方法，代表如果資料存在於 isin 方法內的比對對象，那就傳回 True。
+ 如果比對對象是一個 dictionary：
  + 首先判斷對應的 index 是否存在。
  + 如果存在，接著判斷對應的位置是否存在。
  + 如果不存在，那麼全部為 False。
+ 如果是一個 DataFrame：
  + 首先就是 column 名稱要存在。
  + 原資料行列位置和對應的行列位置一一匹配，才返回 TRUE。
+ 若要設計一個 not in 的分析語法，請在函數方法前面加個 ~ 即可。

```python
import  pandas as pd
data_url='http://bit.ly/2cLzoxH'
gap1 = pd.read_csv(data_url)
print(gap1)
select = gap1["year"].isin([1997,2007])
myData = gap1[select]
print(myData)
gap2 = gap1.query("year==2007 or year == 1997")
print(gap2)
print("="*20)
gap2 = gap1.query("year == 1997 and country=='Afghanistan'")
print(gap2)
gap3 = gap1.query("year == 2007 or (year == 1997 and country == 'Afghanistan')")
print(gap3)
```

## 處理日期資訊
+ 可使用 pandas 的 date_range( ) 函數，可以建立日期資料。
+ 默認情況下，範圍的週期單位為天。
+ 若要表達出工作天 (不包含周六周日)，則請選擇 bdate_range( ) 方法。
+ date_range 或 bdate_range( ) 內的參數：
  + ('起始日期', periods=天數)
  + periods 由開始日取幾天
  + ('起始日期','結束日期')
  + 可加入另外一個參數：freq 代
  + 
 






https://pandas.pydata.org/docs/user_guide/timeseries.html表範圍的週期單位

```python
import pandas as pd
a = pd.date_range("2026/03/11",periods=10)
print(a)
a = pd.date_range("2026/03/11",periods=10,freq="W")
print(a)
a = pd.date_range("2026/03/11",periods=10,freq="W-MON")
print(a)
```

```python
import pandas as pd
from datetime import date
start = date(2026,3,11)
end = date(2026,10,20)
a = pd.date_range(start,end)
print(a)
b = pd.date_range(start,end,freq="W")
print(b)
c = pd.date_range(start,end,freq="ME")
print(c)
```
